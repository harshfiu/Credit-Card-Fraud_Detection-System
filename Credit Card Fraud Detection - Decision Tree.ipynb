```python
# ==============================================
# ROC CURVE - DECISION TREE (UNDERSAMPLED)
# ==============================================

# --- Imports ---
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import RocCurveDisplay, precision_score, recall_score, f1_score, accuracy_score
from imblearn.under_sampling import RandomUnderSampler

# --- Load Dataset ---
df = pd.read_csv('creditcard.csv')

# --- Prepare Data ---
X = df.drop('Class', axis=1)
y = df['Class']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# --- Undersample Data ---
rus = RandomUnderSampler(random_state=42)
X_res, y_res = rus.fit_resample(X_scaled, y)
print(f"✅ Data after undersampling: {y_res.value_counts().to_dict()}")

# --- Train-Test Split ---
X_train, X_test, y_train, y_test = train_test_split(
    X_res, y_res, test_size=0.2, random_state=42, stratify=y_res
)

# --- Train Model ---
model = DecisionTreeClassifier(max_depth=8, random_state=42)
model.fit(X_train, y_train)

# --- Predictions ---
y_pred = model.predict(X_test)
test_acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred)
rec = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print(f"\n✅ Decision Tree ROC Curve (Undersampled) Ready")
print(f"Accuracy: {test_acc:.4f} | Precision: {prec:.4f} | Recall: {rec:.4f} | F1: {f1:.4f}")

# --- Plot ROC Curve ---
plt.figure(figsize=(6,5))
RocCurveDisplay.from_estimator(model, X_test, y_test)
plt.title('ROC Curve - Decision Tree (Undersampled)')
plt.tight_layout()
plt.show()

```

```python
# ==============================================
# CONFUSION MATRIX - DECISION TREE (UNDERSAMPLED)
# ==============================================

# --- Imports ---
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score
from imblearn.under_sampling import RandomUnderSampler

# --- Load Dataset ---
df = pd.read_csv('creditcard.csv')

# --- Prepare Data ---
X = df.drop('Class', axis=1)
y = df['Class']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# --- Undersample Data ---
rus = RandomUnderSampler(random_state=42)
X_res, y_res = rus.fit_resample(X_scaled, y)
print(f"✅ Data after undersampling: {y_res.value_counts().to_dict()}")

# --- Train-Test Split ---
X_train, X_test, y_train, y_test = train_test_split(
    X_res, y_res, test_size=0.2, random_state=42, stratify=y_res
)

# --- Train Model ---
model = DecisionTreeClassifier(max_depth=8, random_state=42)
model.fit(X_train, y_train)

# --- Predictions & Metrics ---
y_pred = model.predict(X_test)
train_acc = model.score(X_train, y_train)
test_acc = model.score(X_test, y_test)
prec = precision_score(y_test, y_pred)
rec = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print(f"\n✅ Decision Tree (Undersampled) Results:")
print(f"Train Accuracy: {train_acc:.4f} | Test Accuracy: {test_acc:.4f}")
print(f"Precision: {prec:.4f} | Recall: {rec:.4f} | F1: {f1:.4f}")

# --- Confusion Matrix Plot ---
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(5,4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Greens')
plt.title('Confusion Matrix - Decision Tree (Undersampled)')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.tight_layout()
plt.show()
```

```python
# ==============================================
# PRECISION–RECALL CURVE - DECISION TREE (UNDERSAMPLED)
# ==============================================

# --- Imports ---
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import PrecisionRecallDisplay, precision_score, recall_score, f1_score, accuracy_score
from imblearn.under_sampling import RandomUnderSampler

# --- Load Dataset ---
df = pd.read_csv('creditcard.csv')

# --- Prepare Data ---
X = df.drop('Class', axis=1)
y = df['Class']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# --- Undersample Data ---
rus = RandomUnderSampler(random_state=42)
X_res, y_res = rus.fit_resample(X_scaled, y)
print(f"✅ Data after undersampling: {y_res.value_counts().to_dict()}")

# --- Train-Test Split ---
X_train, X_test, y_train, y_test = train_test_split(
    X_res, y_res, test_size=0.2, random_state=42, stratify=y_res
)

# --- Train Model ---
model = DecisionTreeClassifier(max_depth=8, random_state=42)
model.fit(X_train, y_train)

# --- Predictions ---
y_pred = model.predict(X_test)
test_acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred)
rec = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print(f"\n✅ Decision Tree Precision–Recall Curve (Undersampled) Ready")
print(f"Accuracy: {test_acc:.4f} | Precision: {prec:.4f} | Recall: {rec:.4f} | F1: {f1:.4f}")

# --- Plot Precision–Recall Curve ---
plt.figure(figsize=(6,5))
PrecisionRecallDisplay.from_estimator(model, X_test, y_test)
plt.title('Precision–Recall Curve - Decision Tree (Undersampled)')
plt.tight_layout()
plt.show()
```
