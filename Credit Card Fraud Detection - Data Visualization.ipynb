```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Display plots inline (for Jupyter)
%matplotlib inline

# -------------------------
# 1️⃣ Load Dataset
# -------------------------
df = pd.read_csv('creditcard.csv')
print("✅ Dataset loaded successfully!")
print("Shape:", df.shape)

# Use 10% sample if dataset is large to avoid lag
if len(df) > 50000:
    df_sample = df.sample(frac=0.1, random_state=42)
    print(f"🔹 Using 10% sample of data ({len(df_sample)} rows) for faster plotting.")
else:
    df_sample = df

# -------------------------
# 2️⃣ Class Distribution (Bar Plot)
# -------------------------
plt.figure(figsize=(5,4))
sns.countplot(x='Class', data=df_sample, palette='Set2')
plt.title('Class Distribution (0 = Non-Fraud, 1 = Fraud)')
plt.xlabel('Transaction Class')
plt.ylabel('Count')
plt.show()

# -------------------------
# 3️⃣ Fraud Ratio (Pie Chart)
# -------------------------
fraud = df_sample[df_sample['Class'] == 1]
non_fraud = df_sample[df_sample['Class'] == 0]

sizes = [len(non_fraud), len(fraud)]
labels = ['Non-Fraud', 'Fraud']
colors = ['#99ff99','#ff9999']

plt.figure(figsize=(5,5))
plt.pie(sizes, labels=labels, autopct='%1.2f%%', startangle=90, colors=colors)
plt.title('Fraud vs Non-Fraud Transaction Ratio')
plt.show()

# -------------------------
# 4️⃣ Amount vs Time Trend (Line Chart)
# -------------------------
plt.figure(figsize=(8,4))
plt.plot(df_sample['Time'], df_sample['Amount'], alpha=0.4, linewidth=0.7)
plt.title('Transaction Amounts Over Time')
plt.xlabel('Time (seconds)')
plt.ylabel('Amount ($)')
plt.grid(alpha=0.3)
plt.show()

# -------------------------
# 5️⃣ Histogram of Transaction Amounts
# -------------------------
plt.figure(figsize=(6,4))
sns.histplot(df_sample['Amount'], bins=50, kde=True, color='skyblue')
plt.title('Transaction Amount Distribution')
plt.xlabel('Amount ($)')
plt.ylabel('Frequency')
plt.show()

# -------------------------
# 6️⃣ Average Amount per Class (Bar Chart)
# -------------------------
avg_amount = df_sample.groupby('Class')['Amount'].mean()
avg_amount.plot(kind='bar', color=['#4c72b0', '#d62728'])
plt.title('Average Transaction Amount by Class')
plt.ylabel('Average Amount ($)')
plt.xticks([0,1], ['Non-Fraud', 'Fraud'], rotation=0)
plt.show()

# -------------------------
# 7️⃣ Feature Correlation Heatmap
# -------------------------
plt.figure(figsize=(10,8))
corr = df_sample.corr()
sns.heatmap(corr, cmap='coolwarm', center=0, linewidths=0.5)
plt.title('Feature Correlation Heatmap (Optimized)')
plt.show()

print("✅ All charts generated successfully!")

plt.title('Average Transaction Amount by Class')
plt.ylabel('Average Amount ($)')
plt.xticks([0,1], ['Non-Fraud', 'Fraud'], rotation=0)
plt.show()

# -------------------------
# 7️⃣ Feature Correlation Heatmap
# -------------------------
plt.figure(figsize=(10,8))
corr = df_sample.corr()
sns.heatmap(corr, cmap='coolwarm', center=0, linewidths=0.5)
plt.title('Feature Correlation Heatmap (Optimized)')
plt.show()

print("✅ All charts generated successfully!")
```
